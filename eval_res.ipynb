### dataset 1

```python
import matplotlib.pyplot as plt
import pandas as pd
df_m1 = eval_res['1_pretrain_large']
df_m2 = eval_res['1_scratch_large']
df_m3 = eval_res['1_pretrain_small']
df_m4 = eval_res['1_scratch_small']
baseline = df_m.iloc[0,:]

# 创建绘图
fig, axes = plt.subplots(nrows=4, ncols=3, figsize=(20, 16))  # 4x3 子图布局
axes = axes.flatten()  # 展平为一维数组以便于索引

# 设置整张图的标题
fig.suptitle("Model Performance Comparison Across Metrics on dataset 1", fontsize=20, fontweight='bold')

# 遍历每个指标
for i, metric in enumerate(df_m1.columns):
    ax = axes[i]
    ax.plot(df_m1[metric], label="pretrain_large")#, marker='o')
    ax.plot(df_m2[metric], label="scratch_large")#, marker='s')
    ax.plot(df_m3[metric], label="pretrain_small")#, marker='^')
    ax.plot(df_m4[metric], label="scratch_small")#, marker='d') 

    # 绘制基线模型的水平线
    baseline_value = baseline[metric]#.values[0]  # 获取基线值
    ax.axhline(y=baseline_value, color='red', linestyle='--', linewidth=1.5, label="Baseline")
    
    ax.set_title(metric)  # 设置子图标题
    ax.set_xlabel("Training Steps")  # 设置 x 轴标签
    ax.set_ylabel("Value")  # 设置 y 轴标签
    ax.set_xticks([])

    # 隐藏图例（仅在第 12 个子图中显示）
    ax.legend().set_visible(False)

# 在第 12 个子图位置绘制集中图例
legend_ax = axes[-1]  # 第 12 个子图
legend_ax.axis('off')  # 关闭坐标轴
handles, labels = axes[0].get_legend_handles_labels()  # 获取第一个子图的图例信息
legend_ax.legend(handles, labels, loc='center', fontsize=12, frameon=True)  # 绘制集中图例

# 隐藏多余的子图
for j in range(len(df_m1.columns), len(axes) - 1):  # 跳过第 12 个子图
    axes[j].axis('off')

# 调整布局以避免重叠
plt.tight_layout(rect=[0, 0, 1, 0.95])  # 为主图标题留出空间

# 显示图像
plt.show()
```
