### dataset 1

```python
import matplotlib.pyplot as plt
import pandas as pd
df_m1 = eval_res['1_pretrain_large']
df_m2 = eval_res['1_scratch_large']
df_m3 = eval_res['1_pretrain_small']
df_m4 = eval_res['1_scratch_small']
baseline = df_m.iloc[0,:]

# 创建绘图
fig, axes = plt.subplots(nrows=4, ncols=3, figsize=(20, 16))  # 4x3 子图布局
axes = axes.flatten()  # 展平为一维数组以便于索引

# 设置整张图的标题
fig.suptitle("Model Performance Comparison Across Metrics on dataset 1", fontsize=20, fontweight='bold')

# 遍历每个指标
for i, metric in enumerate(df_m1.columns):
    ax = axes[i]
    ax.plot(df_m1[metric], label="pretrain_large")#, marker='o')
    ax.plot(df_m2[metric], label="scratch_large")#, marker='s')
    ax.plot(df_m3[metric], label="pretrain_small")#, marker='^')
    ax.plot(df_m4[metric], label="scratch_small")#, marker='d') 

    # 绘制基线模型的水平线
    baseline_value = baseline[metric]#.values[0]  # 获取基线值
    ax.axhline(y=baseline_value, color='red', linestyle='--', linewidth=1.5, label="Baseline")
    
    ax.set_title(metric)  # 设置子图标题
    ax.set_xlabel("Training Steps")  # 设置 x 轴标签
    ax.set_ylabel("Value")  # 设置 y 轴标签
    ax.set_xticks([])

    # 隐藏图例（仅在第 12 个子图中显示）
    ax.legend().set_visible(False)

# 在第 12 个子图位置绘制集中图例
legend_ax = axes[-1]  # 第 12 个子图
legend_ax.axis('off')  # 关闭坐标轴
handles, labels = axes[0].get_legend_handles_labels()  # 获取第一个子图的图例信息
legend_ax.legend(handles, labels, loc='center', fontsize=12, frameon=True)  # 绘制集中图例

# 隐藏多余的子图
for j in range(len(df_m1.columns), len(axes) - 1):  # 跳过第 12 个子图
    axes[j].axis('off')

# 调整布局以避免重叠
plt.tight_layout(rect=[0, 0, 1, 0.95])  # 为主图标题留出空间

# 显示图像
plt.show()
```

### 柱状图展示

#### dataset 1

```python
from matplotlib.cm import get_cmap

## dataset 1 result 
metric_name = res_1_p.columns

sr_pre_l = res_1_p.loc[10000,:]
sr_pre_s = small_1_p.loc[20000,:]
sr_scr_l = res_1_s.loc[10000,:]
sr_scr_s = small_1_s.loc[20000,:]

mvi = df_m.iloc[0,:]
cob = df_cob.iloc[0,:]
scb = df_scb.iloc[0,:]

df = pd.DataFrame([sr_pre_l, sr_pre_s, sr_scr_l, sr_scr_s, mvi,cob, scb], 
                  index=['sr_pre_large', 'sr_pre_small','sr_scr_large','sr_scr_small','mvi', 'cobolt', 'scbutterfly'], columns=metric_name)

df = df.T

# 定义每个方法的颜色
tab10_cmap = get_cmap('tab10')
colors = [tab10_cmap(i) for i in range(len(df.columns))]

# 创建3x4的子图网格
fig, axes = plt.subplots(3, 4, figsize=(15, 10))  # 调整figsize以适应子图大小
axes = axes.flatten()  # 将二维数组展平为一维，方便遍历

# 遍历每个评价指标并绘制直方图
for i, metric in enumerate(df.index):
    ax = axes[i]
    df.loc[metric].plot(kind='bar', ax=ax, color=colors)  # 应用定义的颜色
    ax.set_title(metric, fontsize=10)  # 设置子图标题为评价指标名称
    ax.set_xticks(range(len(df.columns)))  # 设置x轴刻度
    ax.set_xticklabels([])  # 不显示x轴标签
    ax.set_ylabel("Score", fontsize=8)
    ax.tick_params(axis='y', labelsize=8)

# 如果子图数量多于指标数量，隐藏多余的子图
for j in range(len(df.index), len(axes) - 1):  # 留出最后一个子图用于图例
    axes[j].axis('off')

# 在第12张子图的位置添加图例说明
legend_ax = axes[-1]  # 最后一个子图
legend_ax.axis('off')  # 关闭坐标轴
legend_ax.legend(
    handles=[
        plt.Rectangle((0, 0), 1, 1, color=colors[0], label='sr_pre_large'),
        plt.Rectangle((0, 0), 1, 1, color=colors[1], label='sr_pre_small'),
        plt.Rectangle((0, 0), 1, 1, color=colors[2], label='sr_scr_large'),
        plt.Rectangle((0, 0), 1, 1, color=colors[3], label='sr_scr_small'),
        plt.Rectangle((0, 0), 1, 1, color=colors[4], label='mvi'),
        plt.Rectangle((0, 0), 1, 1, color=colors[5], label='cobolt'),
        plt.Rectangle((0, 0), 1, 1, color=colors[6], label='scbutterfly')
    ],
    loc='center',
    fontsize=10,
    frameon=False,
    title="Method Legend",
    title_fontsize=12
)

# 调整布局，避免子图重叠
plt.tight_layout()
plt.show()
```
